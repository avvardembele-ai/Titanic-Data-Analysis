```python
def clean_titanic(df_raw:pd.DataFrame)-> pd.DataFrame:
    df = df_raw.copy()
    #پاک کردن ستون های نامرتبت یا اضافی
    df = df.drop(columns=["Cabin","Ticket","PassengerId"])
    #پر کردن خانه های خالی 
    df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
    df["Age"] = df["Age"].fillna(df["Age"].median())
    #feature engineering 
    #ویژگی های جدید بسازیم
    df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
    df["IsAlone"] = (df["FamilySize"]==1).astype(int)
    
    #پاکسازی متن ها
    df["Title"] = df["Name"].str.extract(r",\s*([^.]*)\.",expand=False).str.strip()
    # , یعنی بعد از ویرگول
    # \s* ممکنه چندتا فاصله باشه
    # ([^.]*) هرچی غیر از نقطه رو بگیر
    # \. تا قبل از نقطه بعدی
    #expand یعنی خروجی یک ستون ساده و معمولی باشه
    #strip فاصله های اول و آخر رو حذف می کنه
    df["Title"] = df["Title"].replace(
        ["Lady","Countess","Capt","Col","Don","Dr","Major","Rev","Sir","Jonkheer","Dona"],
        "Rare"
    )
    df["Title"]=df["Title"].replace(["Mlle","Ms"],"Miss")
    df["Title"]=df["Title"].replace(["Mme"],"Mrs") 
    df = df.drop(columns="Name")
    
    return df
```

```python
df_clean = clean_titanic(df)
df_clean.head(3)
```

```python
df_clean.info()
```

```python
col = [c for c in df_clean.columns if df_clean[c].dtype == "object" ]
df_ready = pd.get_dummies(df_clean,columns=col,drop_first=True)
df_ready =df_ready.astype(int)
df_ready
```

```python
df_ready.to_csv("clean_titanic.csv",index=False)
print("فایل شما با موفقیت ذخیره شد")
```
